## Intelligent Document Analysis & Question Answering System

### 1. Project Overview | نظرة عامة

**English:**

AI Document Intelligence is an end-to-end AI application designed to process, understand, and analyze documents using Large Language Models (LLMs) and Retrieval-Augmented Generation (RAG).

The system allows users to upload PDF, DOCX, and TXT documents, ask questions in natural language, and receive context-aware answers supported by references to the uploaded documents.

**العربية:**

مشروع ذكاء اصطناعي متكامل يهدف إلى قراءة المستندات وتحليلها وفهم محتواها باستخدام نماذج اللغة الكبيرة (LLMs) وتقنية الاسترجاع المعزز بالتوليد (RAG).

يسمح النظام للمستخدم برفع ملفات PDF وDOCX وTXT، وطرح أسئلة باللغة الطبيعية، والحصول على إجابات مستندة إلى محتوى الملفات مع إظهار المصادر المستخدمة.

```python
import numpy as np

from app.ingestion.document_loader import load_document
from app.rag.chunking import split_documents
from app.rag.embeddings import EmbeddingService


# 1 — Load document
documents = load_document(
    "data/sample_documents/employee_handbook.docx"
)

# 2 — Split into chunks
chunks = split_documents(
    documents,
    chunk_size=150,
    chunk_overlap=30,
)

# 3 — Extract chunk texts
texts = [chunk["text"] for chunk in chunks]

# 4 — Load embedding model
embedding_service = EmbeddingService()

# 5 — Generate embeddings
embeddings = embedding_service.encode_texts(texts)

print("Total Chunks:", len(chunks))
print("Embeddings Shape:", embeddings.shape)

# 6 — Encode user question
question = "How many annual leave days do employees receive?"

query_embedding = embedding_service.encode_query(question)

print("Query Embedding Shape:", query_embedding.shape)

# 7 — Calculate cosine similarity
similarities = embeddings @ query_embedding

best_index = int(np.argmax(similarities))

print("\nQuestion:", question)
print("Best Matching Chunk:", best_index + 1)
print("Similarity Score:", round(float(similarities[best_index]), 4))
print("Retrieved Text:", chunks[best_index]["text"])
print("Source:", chunks[best_index]["source"])
```

```python
from app.ingestion.document_loader import load_document
from app.rag.chunking import split_documents
from app.rag.embeddings import EmbeddingService
from app.rag.vector_store import VectorStore


# 1 — Load documents
documents = []

for file_path in [
    "data/sample_documents/company_policy.txt",
    "data/sample_documents/employee_handbook.docx",
    "data/sample_documents/company_security_policy.pdf",
]:
    documents.extend(load_document(file_path))


# 2 — Split documents
chunks = split_documents(
    documents,
    chunk_size=150,
    chunk_overlap=30,
)


# 3 — Initialize embedding service
embedding_service = EmbeddingService()


# 4 — Build FAISS index
vector_store = VectorStore(embedding_service)
vector_store.add_documents(chunks)


# 5 — Search
question = "How many annual leave days do employees receive?"

results = vector_store.search(
    query=question,
    top_k=3,
)

print("Total Indexed Chunks:", vector_store.index.ntotal)
print("\nQuestion:", question)

for rank, result in enumerate(results, start=1):
    print(f"\nResult {rank}")
    print("Score:", round(result["score"], 4))
    print("Source:", result["source"])
    print("Page:", result["page"])
    print("Text:", result["text"])
    print("-" * 50)


# 6 — Save index
vector_store.save("data/faiss_index")
print("\nFAISS index saved successfully.")


# 7 — Reload index
reloaded_store = VectorStore(embedding_service)
reloaded_store.load("data/faiss_index")

reloaded_results = reloaded_store.search(question, top_k=1)

print("\nReload Test:", "PASS" if reloaded_results else "FAIL")
print("Best Retrieved Text:", reloaded_results[0]["text"])
```

```python
# ============================================================
# PROJECT 01 — AI DOCUMENT INTELLIGENCE
# STEP 05 — RAG PIPELINE
# TEST 01 — Grounded Question Answering
#
# PURPOSE:
# 1. Load the previously saved FAISS index.
# 2. Connect FAISS to the local Ollama model.
# 3. Ask a question about employee annual leave.
# 4. Display the generated answer and retrieved sources.
# ============================================================

from app.rag.embeddings import EmbeddingService
from app.rag.vector_store import VectorStore
from app.rag.pipeline import RAGPipeline


# 1 — Initialize embedding model
embedding_service = EmbeddingService()


# 2 — Load existing FAISS index
vector_store = VectorStore(embedding_service)
vector_store.load("data/faiss_index")


# 3 — Initialize RAG pipeline
rag_pipeline = RAGPipeline(
    vector_store=vector_store,
    model_name="qwen2.5:0.5b",
)


# 4 — Ask question
question = "How many annual leave days do employees receive?"

result = rag_pipeline.answer_question(
    question=question,
    top_k=3,
)


# 5 — Display answer
print("QUESTION:", result["question"])
print("\nANSWER:", result["answer"])

print("\nSOURCES:")

for source in result["sources"]:
    print(
        f"- {source['source']} | "
        f"Page: {source['page']} | "
        f"Score: {source['score']}"
    )
```
